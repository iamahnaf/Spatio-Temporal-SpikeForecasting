```python
num_only=df.drop(columns=['wd','station'])
num_only
```

```python
# Get correlation matrix
corr = num_only.corr()

# Find columns with strong correlation (|correlation| > 0.50)
corr_colums = []

for first_position in range(len(corr.columns)):
    for second_position in range(first_position + 1, len(corr.columns)):
        first_column = corr.columns[first_position]
        second_column = corr.columns[second_position]
        correlation_value = corr.loc[first_column, second_column]

        # If correlation is strong (positive or negative)
        if abs(correlation_value) > 0.50:
            if first_column not in corr_colums:
                corr_colums.append(first_column)
            if second_column not in corr_colums:
                corr_colums.append(second_column)

print("Columns with strong correlations (|r| > 0.50):")
print(corr_colums)
```

```python
plt.figure(figsize=(8, 5))
wd_count= df['wd'].value_counts().sort_index().reset_index()
wd_count.columns = ['wd','count']
sns.barplot(data=wd_count,x='wd',y='count')
plt.title("Wind direction frequency (all stations)")
plt.tight_layout()
```

```python
# Create the datetime column used for monthly grouping
df["datetime"] = pd.to_datetime(
    df[["year", "month", "day", "hour"]]
)

# Mark PM2.5 values above 150 as spike events
df["is_spike"] = (df["PM2.5"] > 150).astype(int)
```

```python
spikes_over_time = df[df["is_spike"] == 1].groupby(
    df["datetime"].dt.to_period("M"))["is_spike"].count()
plt.figure(figsize=(8, 6))
spikes_over_time.plot(kind="bar")
plt.title("Spike events per month (all stations combined)")
plt.tight_layout()
```

```python
from pandas.plotting import autocorrelation_plot
dhaka = df[df["station"] == "Dhaka"].set_index("datetime")["PM2.5"]
plt.figure(figsize=(10, 5))
autocorrelation_plot(dhaka.iloc[:2000])  # subset for speed
plt.title("Autocorrelation of PM2.5 - Dhaka (first 2000 hrs)")
plt.tight_layout()
```
